```python
class Laptop:
    def __init__(self, price):
        self.__price = price

    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
        else:
            print("Price must be positive!")

    def get_price(self):
        return self.__price

l1 = Laptop(50000)

print("Original Price:", l1.get_price())

l1.set_price(55000)
print("Updated Price:", l1.get_price())      
```

##Question 2:


```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposited:", amount)
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance or invalid amount.")

    def check_balance(self):
        return self.__balance


# Example usage
acc = BankAccount(1000)
print("Initial Balance:", acc.check_balance())

acc.deposit(500)
print("Balance after deposit:", acc.check_balance())

acc.withdraw(300)
print("Balance after withdrawal:", acc.check_balance())


```

##Question 3:

```python
class Movie:
    def __init__(self, rating):
        self.__rating = None
        self.set_rating(rating)

    def set_rating(self, rating):
        if 1 <= rating <= 10:
            self.__rating = rating
            print("Rating set to:", rating)
        else:
            print("Rating must be between 1 and 10.")

    def get_rating(self):
        return self.__rating


# Example usage
m1 = Movie(8)
print("Current Rating:", m1.get_rating())

m1.set_rating(11)   # Invalid
m1.set_rating(9)    # Valid
print("Updated Rating:", m1.get_rating())

```

```python
class Book:
    def __init__(self, title, author):
        self.__title = title
        self.__author = author

    def get_title(self):
        return self.__title

    def get_author(self):
        return self.__author


# Example usage
b1 = Book("Python Basics", "John Smith")
print("Title:", b1.get_title())
print("Author:", b1.get_author())
```

```python
class Car:
    def __init__(self, speed=0):
        self.__speed = speed

    def increase_speed(self, amount):
        if amount > 0:
            self.__speed += amount
            print("Speed increased by", amount)

    def decrease_speed(self, amount):
        if amount > 0:
            self.__speed -= amount
            if self.__speed < 0:
                self.__speed = 0
            print("Speed decreased by", amount)

    def get_current_speed(self):
        return self.__speed


# Example usage
c1 = Car()
print("Initial Speed:", c1.get_current_speed())

c1.increase_speed(60)
print("Speed now:", c1.get_current_speed())

c1.decrease_speed(80)  # Should not go below 0
print("Final Speed:", c1.get_current_speed())
```
